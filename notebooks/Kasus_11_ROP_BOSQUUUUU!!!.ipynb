```python
# ============================================================
# ROP OPTIMIZATION - TWO DIFFERENT DRILLING / WELL
#
# CSV 2 = DEVELOPMENT / TRAINING WELL
# CSV 1 = EXTERNAL TEST WELL
#
# Alur:
# CSV 2
#   -> Temporal Split berdasarkan Hole Depth
#   -> Optuna tuning
#   -> Evaluasi internal
#   -> Pilih model terbaik
#   -> Retrain model terbaik dengan CSV 2 FULL
#   -> External testing menggunakan CSV 1
#
# Models:
# Random Forest
# XGBoost
# CatBoost
# ============================================================


# ============================================================
# 1. INSTALL LIBRARY
# ============================================================
!pip install -q optuna xgboost catboost openpyxl


# ============================================================
# 2. IMPORT LIBRARY
# ============================================================
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import optuna
import joblib

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from xgboost import XGBRegressor
from catboost import CatBoostRegressor


# ============================================================
# 3. PATH FILE
# ============================================================
# GANTI SESUAI NAMA FILE DI /content/

CSV2_PATH = "/content/well2.csv"
CSV1_PATH = "/content/well1.csv"


# ============================================================
# 4. LOAD CSV 2 DAN CSV 1
# ============================================================
df2 = pd.read_csv(CSV2_PATH)
df1 = pd.read_csv(CSV1_PATH)

print("=" * 80)
print("INFORMASI DATA")
print("=" * 80)

print("\nCSV 2 - Development / Training Well")
print("Shape :", df2.shape)

print("\nCSV 1 - External Test Well")
print("Shape :", df1.shape)


# ============================================================
# 5. STANDARDISASI NAMA KOLOM
# ============================================================
def standardize_columns(df):
    df = df.copy()

    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_", regex=False)
    )

    return df


df2 = standardize_columns(df2)
df1 = standardize_columns(df1)


print("\nKolom CSV 2:")
print(df2.columns.tolist())

print("\nKolom CSV 1:")
print(df1.columns.tolist())


# ============================================================
# 6. DEFINISI TARGET DAN INDEX
# ============================================================
TARGET = "Rate_Of_Penetration"
DEPTH = "Hole_Depth"


# Pastikan kolom tersedia
for name, df in [("CSV 2", df2), ("CSV 1", df1)]:

    if TARGET not in df.columns:
        raise ValueError(
            f"Kolom '{TARGET}' tidak ditemukan pada {name}."
        )

    if DEPTH not in df.columns:
        raise ValueError(
            f"Kolom '{DEPTH}' tidak ditemukan pada {name}."
        )


# ============================================================
# 7. KONVERSI DATA KE NUMERIC
# ============================================================
for df in [df2, df1]:

    for col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )


# ============================================================
# 8. DROP MISSING VALUE PADA DEPTH DAN TARGET
# ============================================================
df2 = df2.dropna(
    subset=[DEPTH, TARGET]
).copy()

df1 = df1.dropna(
    subset=[DEPTH, TARGET]
).copy()


# ============================================================
# 9. CEK ROP = 0 DAN ROP NEGATIF
# ============================================================
print("\n" + "=" * 80)
print("PEMERIKSAAN ROP")
print("=" * 80)

print("\nCSV 2:")
print(
    "ROP = 0      :",
    (df2[TARGET] == 0).sum(),
    f"({(df2[TARGET] == 0).mean()*100:.2f}%)"
)

print(
    "ROP < 0      :",
    (df2[TARGET] < 0).sum()
)

print("\nCSV 1:")
print(
    "ROP = 0      :",
    (df1[TARGET] == 0).sum(),
    f"({(df1[TARGET] == 0).mean()*100:.2f}%)"
)

print(
    "ROP < 0      :",
    (df1[TARGET] < 0).sum()
)


# ============================================================
# 10. FILTER ROP NEGATIF
# ============================================================
# ROP = 0 TETAP DIPERTAHANKAN.
#
# Jika pada dataset ternyata ROP negatif memiliki makna
# operasional tertentu, bagian ini perlu ditinjau kembali.

df2 = df2[df2[TARGET] >= 0].copy()
df1 = df1[df1[TARGET] >= 0].copy()


# ============================================================
# 11. SORT BERDASARKAN HOLE DEPTH
# ============================================================
df2 = (
    df2
    .sort_values(DEPTH)
    .reset_index(drop=True)
)

df1 = (
    df1
    .sort_values(DEPTH)
    .reset_index(drop=True)
)


# ============================================================
# 12. HOLE DEPTH MENJADI INDEX
# ============================================================
df2 = df2.set_index(DEPTH)
df1 = df1.set_index(DEPTH)


print("\n" + "=" * 80)
print("RENTANG HOLE DEPTH")
print("=" * 80)

print(
    f"CSV 2 : {df2.index.min()} -> {df2.index.max()}"
)

print(
    f"CSV 1 : {df1.index.min()} -> {df1.index.max()}"
)


# ============================================================
# 13. SAMAKAN FEATURE ANTARA CSV 2 DAN CSV 1
# ============================================================
# Hanya fitur yang tersedia di KEDUA pengeboran digunakan.
#
# Hole Depth = index
# Rate Of Penetration = target

common_features = [
    col for col in df2.columns
    if col in df1.columns
    and col != TARGET
]


print("\n" + "=" * 80)
print("FEATURE YANG DIGUNAKAN")
print("=" * 80)

print("Jumlah feature :", len(common_features))

for i, col in enumerate(common_features, 1):
    print(f"{i:02d}. {col}")


# ============================================================
# 14. BUAT X DAN y
# ============================================================
X2 = df2[common_features].copy()
y2 = df2[TARGET].copy()

X1 = df1[common_features].copy()
y1 = df1[TARGET].copy()


# ============================================================
# 15. HANDLE INFINITE VALUE
# ============================================================
X2 = X2.replace(
    [np.inf, -np.inf],
    np.nan
)

X1 = X1.replace(
    [np.inf, -np.inf],
    np.nan
)


# Jika terdapat missing feature, gunakan median CSV 2.
# Median hanya dihitung dari data training/development.
#
# Ini mencegah informasi CSV 1 masuk ke proses training.

train_medians = X2.median()

X2 = X2.fillna(train_medians)
X1 = X1.fillna(train_medians)


# ============================================================
# 16. TEMPORAL SPLIT CSV 2
# ============================================================
# 80% depth awal -> training
# 20% depth akhir -> validation
#
# Tidak menggunakan random split.

split_idx = int(len(X2) * 0.80)

X2_train = X2.iloc[:split_idx].copy()
X2_val = X2.iloc[split_idx:].copy()

y2_train = y2.iloc[:split_idx].copy()
y2_val = y2.iloc[split_idx:].copy()


print("\n" + "=" * 80)
print("TEMPORAL SPLIT CSV 2")
print("=" * 80)

print("Training :", X2_train.shape)
print("Validation :", X2_val.shape)

print("\nHole Depth:")
print(
    "Training   :",
    X2_train.index.min(),
    "->",
    X2_train.index.max()
)

print(
    "Validation :",
    X2_val.index.min(),
    "->",
    X2_val.index.max()
)


# ============================================================
# 17. TIME SERIES CROSS VALIDATION
# ============================================================
tscv = TimeSeriesSplit(n_splits=3)


# ============================================================
# 18. FUNGSI RMSE
# ============================================================
def rmse_score(y_true, y_pred):

    return np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )


# ============================================================
# 19. OPTUNA - RANDOM FOREST
# ============================================================
print("\n" + "=" * 80)
print("OPTUNA - RANDOM FOREST")
print("=" * 80)


def objective_rf(trial):

    params = {

        "n_estimators": trial.suggest_int(
            "n_estimators",
            200,
            1000,
            step=100
        ),

        "max_depth": trial.suggest_int(
            "max_depth",
            5,
            50
        ),

        "min_samples_split": trial.suggest_int(
            "min_samples_split",
            2,
            20
        ),

        "min_samples_leaf": trial.suggest_int(
            "min_samples_leaf",
            1,
            10
        ),

        "max_features": trial.suggest_categorical(
            "max_features",
            [
                "sqrt",
                "log2",
                0.3,
                0.5,
                0.7,
                1.0
            ]
        ),

        "bootstrap": trial.suggest_categorical(
            "bootstrap",
            [True, False]
        ),

        "random_state": 42,
        "n_jobs": -1
    }

    model = RandomForestRegressor(**params)

    scores = []

    for train_idx, val_idx in tscv.split(X2_train):

        X_train_fold = X2_train.iloc[train_idx]
        X_val_fold = X2_train.iloc[val_idx]

        y_train_fold = y2_train.iloc[train_idx]
        y_val_fold = y2_train.iloc[val_idx]

        model.fit(
            X_train_fold,
            y_train_fold
        )

        pred = model.predict(
            X_val_fold
        )

        score = rmse_score(
            y_val_fold,
            pred
        )

        scores.append(score)

    return np.mean(scores)


study_rf = optuna.create_study(
    direction="minimize",
    study_name="ROP_RandomForest"
)

study_rf.optimize(
    objective_rf,
    n_trials=30,
    show_progress_bar=True
)

best_rf_params = study_rf.best_params

print("\nBest RF CV RMSE:")
print(study_rf.best_value)

print("\nBest RF Parameters:")
print(best_rf_params)


# ============================================================
# 20. OPTUNA - XGBOOST
# ============================================================
print("\n" + "=" * 80)
print("OPTUNA - XGBOOST")
print("=" * 80)


def objective_xgb(trial):

    params = {

        "n_estimators": trial.suggest_int(
            "n_estimators",
            300,
            1500,
            step=100
        ),

        "max_depth": trial.suggest_int(
            "max_depth",
            3,
            12
        ),

        "learning_rate": trial.suggest_float(
            "learning_rate",
            0.005,
            0.3,
            log=True
        ),

        "subsample": trial.suggest_float(
            "subsample",
            0.5,
            1.0
        ),

        "colsample_bytree": trial.suggest_float(
            "colsample_bytree",
            0.5,
            1.0
        ),

        "min_child_weight": trial.suggest_int(
            "min_child_weight",
            1,
            20
        ),

        "gamma": trial.suggest_float(
            "gamma",
            0,
            5
        ),

        "reg_alpha": trial.suggest_float(
            "reg_alpha",
            1e-8,
            10,
            log=True
        ),

        "reg_lambda": trial.suggest_float(
            "reg_lambda",
            1e-3,
            100,
            log=True
        ),

        "objective": "reg:squarederror",
        "eval_metric": "rmse",
        "random_state": 42,
        "n_jobs": -1
    }

    model = XGBRegressor(**params)

    scores = []

    for train_idx, val_idx in tscv.split(X2_train):

        X_train_fold = X2_train.iloc[train_idx]
        X_val_fold = X2_train.iloc[val_idx]

        y_train_fold = y2_train.iloc[train_idx]
        y_val_fold = y2_train.iloc[val_idx]

        model.fit(
            X_train_fold,
            y_train_fold
        )

        pred = model.predict(
            X_val_fold
        )

        score = rmse_score(
            y_val_fold,
            pred
        )

        scores.append(score)

    return np.mean(scores)


study_xgb = optuna.create_study(
    direction="minimize",
    study_name="ROP_XGBoost"
)

study_xgb.optimize(
    objective_xgb,
    n_trials=40,
    show_progress_bar=True
)

best_xgb_params = study_xgb.best_params

print("\nBest XGBoost CV RMSE:")
print(study_xgb.best_value)

print("\nBest XGBoost Parameters:")
print(best_xgb_params)


# ============================================================
# 21. OPTUNA - CATBOOST
# ============================================================
print("\n" + "=" * 80)
print("OPTUNA - CATBOOST")
print("=" * 80)


def objective_cat(trial):

    params = {

        "iterations": trial.suggest_int(
            "iterations",
            300,
            1500,
            step=100
        ),

        "depth": trial.suggest_int(
            "depth",
            4,
            12
        ),

        "learning_rate": trial.suggest_float(
            "learning_rate",
            0.005,
            0.3,
            log=True
        ),

        "l2_leaf_reg": trial.suggest_float(
            "l2_leaf_reg",
            1,
            100,
            log=True
        ),

        "random_strength": trial.suggest_float(
            "random_strength",
            0,
            5
        ),

        "bagging_temperature": trial.suggest_float(
            "bagging_temperature",
            0,
            5
        ),

        "loss_function": "RMSE",
        "verbose": False,
        "random_seed": 42,
        "thread_count": -1
    }

    model = CatBoostRegressor(**params)

    scores = []

    for train_idx, val_idx in tscv.split(X2_train):

        X_train_fold = X2_train.iloc[train_idx]
        X_val_fold = X2_train.iloc[val_idx]

        y_train_fold = y2_train.iloc[train_idx]
        y_val_fold = y2_train.iloc[val_idx]

        model.fit(
            X_train_fold,
            y_train_fold
        )

        pred = model.predict(
            X_val_fold
        )

        score = rmse_score(
            y_val_fold,
            pred
        )

        scores.append(score)

    return np.mean(scores)


study_cat = optuna.create_study(
    direction="minimize",
    study_name="ROP_CatBoost"
)

study_cat.optimize(
    objective_cat,
    n_trials=40,
    show_progress_bar=True
)

best_cat_params = study_cat.best_params

print("\nBest CatBoost CV RMSE:")
print(study_cat.best_value)

print("\nBest CatBoost Parameters:")
print(best_cat_params)


# ============================================================
# 22. TRAIN MODEL DENGAN PARAMETER TERBAIK
#    PADA TRAINING CSV 2
# ============================================================

rf_dev = RandomForestRegressor(
    **best_rf_params,
    random_state=42,
    n_jobs=-1
)

xgb_dev = XGBRegressor(
    **best_xgb_params,
    objective="reg:squarederror",
    eval_metric="rmse",
    random_state=42,
    n_jobs=-1
)

cat_dev = CatBoostRegressor(
    **best_cat_params,
    loss_function="RMSE",
    verbose=False,
    random_seed=42,
    thread_count=-1
)


rf_dev.fit(X2_train, y2_train)
xgb_dev.fit(X2_train, y2_train)
cat_dev.fit(X2_train, y2_train)


# ============================================================
# 23. PREDIKSI VALIDATION CSV 2
# ============================================================
rf_val_pred = rf_dev.predict(X2_val)
xgb_val_pred = xgb_dev.predict(X2_val)
cat_val_pred = cat_dev.predict(X2_val)


# ============================================================
# 24. EVALUASI CSV 2 VALIDATION
# ============================================================
def evaluate_model(
    model_name,
    y_true,
    y_pred
):

    return {

        "Model": model_name,

        "MAE": mean_absolute_error(
            y_true,
            y_pred
        ),

        "RMSE": np.sqrt(
            mean_squared_error(
                y_true,
                y_pred
            )
        ),

        "R2": r2_score(
            y_true,
            y_pred
        )
    }


validation_results = pd.DataFrame([

    evaluate_model(
        "Random Forest",
        y2_val,
        rf_val_pred
    ),

    evaluate_model(
        "XGBoost",
        y2_val,
        xgb_val_pred
    ),

    evaluate_model(
        "CatBoost",
        y2_val,
        cat_val_pred
    )

])

validation_results = validation_results.sort_values(
    "RMSE"
).reset_index(drop=True)


print("\n" + "=" * 80)
print("HASIL VALIDATION - CSV 2")
print("=" * 80)

display(validation_results)


# ============================================================
# 25. TENTUKAN MODEL TERBAIK BERDASARKAN VALIDATION RMSE
# ============================================================
best_model_name = validation_results.iloc[0]["Model"]

print("\nModel terbaik berdasarkan RMSE validation CSV 2:")
print(best_model_name)


# ============================================================
# 26. RETRAIN MODEL TERBAIK MENGGUNAKAN CSV 2 FULL
# ============================================================
print("\n" + "=" * 80)
print("RETRAIN MODEL TERBAIK DENGAN CSV 2 FULL")
print("=" * 80)


if best_model_name == "Random Forest":

    final_model = RandomForestRegressor(
        **best_rf_params,
        random_state=42,
        n_jobs=-1
    )

elif best_model_name == "XGBoost":

    final_model = XGBRegressor(
        **best_xgb_params,
        objective="reg:squarederror",
        eval_metric="rmse",
        random_state=42,
        n_jobs=-1
    )

elif best_model_name == "CatBoost":

    final_model = CatBoostRegressor(
        **best_cat_params,
        loss_function="RMSE",
        verbose=False,
        random_seed=42,
        thread_count=-1
    )


# Training dengan SELURUH CSV 2
final_model.fit(
    X2,
    y2
)

print("Final model:", best_model_name)
print("Training data:", X2.shape)


# ============================================================
# 27. EXTERNAL TESTING - CSV 1
# ============================================================
external_pred = final_model.predict(X1)


# ============================================================
# 28. EVALUASI EXTERNAL TEST CSV 1
# ============================================================
external_results = pd.DataFrame([

    evaluate_model(
        best_model_name,
        y1,
        external_pred
    )

])


print("\n" + "=" * 80)
print("HASIL EXTERNAL TESTING - CSV 1")
print("=" * 80)

display(external_results)


# ============================================================
# 29. HASIL PREDIKSI CSV 1
# ============================================================
external_prediction_result = pd.DataFrame({

    "Hole_Depth": X1.index,

    "Actual_ROP": y1.values,

    "Predicted_ROP": external_pred

})


print("\n" + "=" * 80)
print("CONTOH HASIL PREDIKSI CSV 1")
print("=" * 80)

display(
    external_prediction_result.head(20)
)


# ============================================================
# 30. GRAFIK ACTUAL VS PREDICTED - CSV 2 VALIDATION
# ============================================================
plt.figure(figsize=(15, 6))

plt.plot(
    X2_val.index,
    y2_val.values,
    label="Actual ROP"
)

plt.plot(
    X2_val.index,
    rf_val_pred,
    label="Random Forest"
)

plt.plot(
    X2_val.index,
    xgb_val_pred,
    label="XGBoost"
)

plt.plot(
    X2_val.index,
    cat_val_pred,
    label="CatBoost"
)

plt.xlabel("Hole Depth")
plt.ylabel("Rate Of Penetration")

plt.title(
    "CSV 2 - Temporal Validation: Actual vs Predicted ROP"
)

plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# 31. GRAFIK ACTUAL VS PREDICTED - CSV 1
# ============================================================
plt.figure(figsize=(15, 6))

plt.plot(
    X1.index,
    y1.values,
    label="Actual ROP"
)

plt.plot(
    X1.index,
    external_pred,
    label=f"Predicted ROP - {best_model_name}"
)

plt.xlabel("Hole Depth")
plt.ylabel("Rate Of Penetration")

plt.title(
    f"CSV 1 - External Test: Actual vs Predicted ROP\n"
    f"Model: {best_model_name}"
)

plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# 32. SCATTER ACTUAL VS PREDICTED - CSV 1
# ============================================================
plt.figure(figsize=(7, 7))

plt.scatter(
    y1,
    external_pred,
    alpha=0.5
)

min_value = min(
    y1.min(),
    external_pred.min()
)

max_value = max(
    y1.max(),
    external_pred.max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.xlabel("Actual ROP")
plt.ylabel("Predicted ROP")

plt.title(
    f"CSV 1 External Test - {best_model_name}"
)

plt.grid(True)

plt.show()


# ============================================================
# 33. FEATURE IMPORTANCE MODEL FINAL
# ============================================================
if best_model_name == "Random Forest":

    importance_values = final_model.feature_importances_

elif best_model_name == "XGBoost":

    importance_values = final_model.feature_importances_

elif best_model_name == "CatBoost":

    importance_values = final_model.get_feature_importance()


feature_importance = pd.DataFrame({

    "Feature": common_features,

    "Importance": importance_values

}).sort_values(
    "Importance",
    ascending=False
).reset_index(drop=True)


print("\n" + "=" * 80)
print("FEATURE IMPORTANCE - FINAL MODEL")
print("=" * 80)

display(feature_importance)


# ============================================================
# 34. PLOT FEATURE IMPORTANCE
# ============================================================
plt.figure(figsize=(10, 7))

top_n = min(
    15,
    len(feature_importance)
)

plt.barh(
    feature_importance["Feature"].head(top_n)[::-1],
    feature_importance["Importance"].head(top_n)[::-1]
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.title(
    f"Top {top_n} Feature Importance - {best_model_name}"
)

plt.grid(axis="x")

plt.show()


# ============================================================
# 35. RESIDUAL EXTERNAL TEST
# ============================================================
external_residual = (
    y1.values -
    external_pred
)


plt.figure(figsize=(15, 5))

plt.scatter(
    X1.index,
    external_residual,
    alpha=0.5
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel("Hole Depth")
plt.ylabel("Residual (Actual - Predicted)")

plt.title(
    f"Residual Analysis - CSV 1 External Test\n"
    f"{best_model_name}"
)

plt.grid(True)

plt.show()


# ============================================================
# 36. SIMPAN MODEL
# ============================================================
joblib.dump(
    final_model,
    "/content/rop_final_model.pkl"
)


# ============================================================
# 37. SIMPAN HASIL VALIDATION
# ============================================================
validation_results.to_excel(
    "/content/rop_csv2_validation_results.xlsx",
    index=False
)


# ============================================================
# 38. SIMPAN HASIL EXTERNAL TEST
# ============================================================
external_results.to_excel(
    "/content/rop_csv1_external_test_results.xlsx",
    index=False
)


# ============================================================
# 39. SIMPAN PREDIKSI CSV 1
# ============================================================
external_prediction_result.to_excel(
    "/content/rop_csv1_predictions.xlsx",
    index=False
)


# ============================================================
# 40. SIMPAN FEATURE IMPORTANCE
# ============================================================
feature_importance.to_excel(
    "/content/rop_final_feature_importance.xlsx",
    index=False
)


# ============================================================
# 41. SIMPAN BEST PARAMETERS
# ============================================================
best_parameters = {

    "Best_Model": best_model_name,

    "Random_Forest": best_rf_params,

    "XGBoost": best_xgb_params,

    "CatBoost": best_cat_params

}

# Simpan sebagai text
with open(
    "/content/rop_best_parameters.txt",
    "w"
) as f:

    f.write(
        str(best_parameters)
    )


# ============================================================
# 42. RINGKASAN AKHIR
# ============================================================
print("\n")
print("=" * 80)
print("RINGKASAN AKHIR")
print("=" * 80)

print("\nDATA:")
print(
    f"CSV 2 (Development) : "
    f"{df2.index.min()} -> {df2.index.max()}"
)

print(
    f"CSV 1 (External Test): "
    f"{df1.index.min()} -> {df1.index.max()}"
)

print("\nMODEL TERBAIK:")
print(best_model_name)

print("\n--- VALIDATION CSV 2 ---")
display(validation_results)

print("\n--- EXTERNAL TEST CSV 1 ---")
display(external_results)

print("\nFile output:")
print("/content/rop_final_model.pkl")
print("/content/rop_csv2_validation_results.xlsx")
print("/content/rop_csv1_external_test_results.xlsx")
print("/content/rop_csv1_predictions.xlsx")
print("/content/rop_final_feature_importance.xlsx")
print("/content/rop_best_parameters.txt")

print("\n" + "=" * 80)
print("SELESAI")
print("=" * 80)
```

```python
# ============================================================
# SIMPAN ULANG FINAL MODEL RF DENGAN JOBLIB
# ============================================================

import joblib
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestRegressor

# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------

CSV2_PATH = "/content/well2.csv"

DEPTH = "Hole_Depth"
TARGET = "Rate_Of_Penetration"

df2 = pd.read_csv(CSV2_PATH)

# Standardisasi nama kolom
df2.columns = (
    df2.columns
    .str.strip()
    .str.replace(" ", "_", regex=False)
)

# Konversi numerik
for col in df2.columns:
    df2[col] = pd.to_numeric(df2[col], errors="coerce")

# Cleaning
df2 = df2.dropna(subset=[DEPTH, TARGET]).copy()
df2 = df2[df2[TARGET] >= 0].copy()
df2 = df2.sort_values(DEPTH).reset_index(drop=True)

print("Ukuran data:", df2.shape)
print("Hole Depth:", df2[DEPTH].min(), "->", df2[DEPTH].max())

# ------------------------------------------------------------
# 2. FEATURES
# ------------------------------------------------------------

FEATURES = [
    col for col in df2.columns
    if col not in [DEPTH, TARGET]
]

X = df2[FEATURES].copy()
y = df2[TARGET].copy()

# Tangani inf
X = X.replace([np.inf, -np.inf], np.nan)

# Median
medians = X.median()

# Isi missing
X = X.fillna(medians)

print("\nJumlah features:", len(FEATURES))
print(FEATURES)

# ------------------------------------------------------------
# 3. BEST PARAMETERS HASIL OPTUNA
# ------------------------------------------------------------

RF_PARAMS = {
    "n_estimators": 900,
    "max_depth": 47,
    "min_samples_split": 5,
    "min_samples_leaf": 7,
    "max_features": 1.0,
    "bootstrap": True,
    "random_state": 42,
    "n_jobs": -1
}

# ------------------------------------------------------------
# 4. TRAIN FINAL RANDOM FOREST
# ------------------------------------------------------------

print("\nTraining Random Forest...")

rf_model = RandomForestRegressor(**RF_PARAMS)

rf_model.fit(X, y)

print("Training selesai.")

# ------------------------------------------------------------
# 5. BUAT PACKAGE
# ------------------------------------------------------------

rf_package = {
    "model": rf_model,
    "features": FEATURES,
    "medians": medians,
    "target": TARGET,
    "depth_column": DEPTH,
    "model_name": "Random Forest",
    "best_parameters": RF_PARAMS
}

# ------------------------------------------------------------
# 6. SIMPAN DENGAN JOBLIB
# ------------------------------------------------------------

RF_PATH = "/content/rop_final_rf.joblib"

joblib.dump(
    rf_package,
    RF_PATH,
    compress=3
)

print("\nModel berhasil disimpan:")
print(RF_PATH)
```
